Importing the sheets to extract samples from every sheet

```python
import pandas as pd
w_m=pd.read_excel("final_output.xlsx", sheet_name="WASHING MACHINES")
fridges=pd.read_excel("final_output.xlsx", sheet_name="FRIDGES")
d_w=pd.read_excel("final_output.xlsx", sheet_name="DISHWASHERS")
hoods=pd.read_excel("final_output.xlsx", sheet_name="HOODS")
kitchens=pd.read_excel("final_output.xlsx", sheet_name="KITCHENS")
```

```python
w_m=w_m.sample(4)
fridges=fridges.sample(4)
d_w=d_w.sample(4)
hoods=hoods.sample(4)
kitchens=kitchens.sample(4)
```

```python
with pd.ExcelWriter("sample_out.xlsx") as writer:
    d_w.to_excel(writer, sheet_name="DISHWASHERS", index=False)
    fridges.to_excel(writer, sheet_name="FRIDGES", index=False)
    w_m.to_excel(writer, sheet_name="WASHING MACHINES", index=False)
    hoods.to_excel(writer, sheet_name="HOODS", index=False)
    kitchens.to_excel(writer, sheet_name="KITCHENS", index=False)
```
